```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df = pd.read_csv('Characters.csv', sep=';')
```

```python
df.sample(7)
```

```python
df.shape
```

```python
df.duplicated().sum()
```

```python
plt.figure(figsize=(6, 4))
sns.countplot(data=df, x='Gender', order=df['Gender'].value_counts().index)
plt.title('Distribution of Gender')
plt.xlabel('Gender')
plt.ylabel('Count')
plt.show()
```

```python
plt.figure(figsize=(8, 4))
sns.countplot(data=df, x='House', order=df['House'].value_counts().index)
plt.title('Distribution of House')
plt.xlabel('House')
plt.ylabel('Count')
plt.xticks(rotation=45)
plt.show()
```

```python
plt.figure(figsize=(9, 4))
sns.countplot(data=df, x='Blood status', order=df['Blood status'].value_counts().index)
plt.title('Distribution of Blood Status')
plt.xlabel('Blood status')
plt.ylabel('Count')
plt.xticks(rotation=45)
plt.show()
```

```python
plt.figure(figsize=(10, 4))
sns.countplot(data=df, x='Species', order=df['Species'].value_counts().index)
plt.title('Distribution of Species')
plt.xlabel('Species')
plt.ylabel('Count')
plt.xticks(rotation=45)
plt.show()
```

```python
plt.figure(figsize=(10, 5))
top_jobs = df['Job'].value_counts().head(10)
sns.barplot(x=top_jobs.values, y=top_jobs.index, palette='viridis')
plt.title('Top 10 Character Jobs / Roles')
plt.xlabel('Count')
plt.ylabel('Job')
plt.show()
```
