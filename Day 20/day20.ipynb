```python
import pandas as pd
import seaborn as sns
```

```python
df = pd.read_csv('train.csv')
```

```python
df.sample(7)
```

Data - has columns which contain multiple values of that feature. The columns of data can either be NUMERICAL or CATEGORICAL
We deal with both types of data differently and separately.
Generally, we start left to right when analysing columns but here we have divided them into the two classes.

# 1. Categorical Data

## a. Countplot

```python
sns.countplot(data=df, x='Survived')
# df['Survived'].value_counts().plot(kind='bar')
```

```python
sns.countplot(data=df, x='Pclass')
```

```python
sns.countplot(df['Embarked'])
```

## b. PieChart

```python
df['Sex'].value_counts().plot(kind='pie', autopct='%.2f')
# use autopct='%.2f' to display percentages in the pie chart
```

# 2. Numerical Data

## a. Histogram

```python
import matplotlib.pyplot as plt
```

```python
plt.hist(df['Age'], bins=8)
```

## Distplot (for Kernel Density Estimate)

```python
sns.kdeplot(df['Age'])
```

```python
sns.histplot(df['Age'], kde=True)
# KDE plot shows the probability density. Area under KDE plot is 1
```

## c. Boxplot



Use when data is noisy (many outliers)

```python
sns.boxplot(df['Fare'])

# The central rectangle (the box) represents the Interquartile Range (IQR), which contains exactly the middle 50% of your data.
# A line drawn inside this box marks the median, showing you the exact middle value of the entire dataset.
# Lines extending outward from the box (the whiskers) represent the acceptable upper and lower boundaries of your data's spread.
# Any data points that fall mathematically outside of these whiskers are plotted as individual dots, immediately flagging them as outliers.
```

```python
df['Age'].min()
```

```python
df['Age'].max()
```

```python
df['Age'].mean()
```

```python
df['Age'].skew()
# >0 means right of y axis
# <0 means left of y axis
```
